# 06 - Fondos Comunes de Inversion (FCI)

Este notebook prueba las funcionalidades de FCI de la API de IOL.

## Funcionalidades:
- Listar fondos disponibles
- Detalle de un fondo especifico
- Tipos de fondos y administradoras
- Suscripcion a fondos
- Rescate de fondos

**ADVERTENCIA:** Las operaciones de suscripcion/rescate estan COMENTADAS.

**Nota:** Requiere credenciales validas de IOL en el archivo `.env`

## Configuracion Inicial

```python
import sys
import os
from datetime import datetime
from dotenv import load_dotenv

sys.path.insert(0, os.path.abspath('../..'))

from pyIol import (
    IOLClient, IOLAPIError,
    FondoComunInversion, FCIDetalle, TipoFondo, AdministradoraFCI, ResultadoFCI
)
print("Librerias importadas correctamente")
print(f"Fecha y hora: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```

```python
load_dotenv('../../.env')
USERNAME = os.getenv('IOL_USERNAME', 'tu_usuario_iol')
PASSWORD = os.getenv('IOL_PASSWORD', 'tu_password_iol')

if USERNAME == "tu_usuario_iol":
    print("ADVERTENCIA: Configura las credenciales en .env")
else:
    print(f"Credenciales configuradas - Usuario: {USERNAME}")
```

```python
try:
    client = IOLClient(USERNAME, PASSWORD)
    print("Cliente IOL creado correctamente")
except Exception as e:
    print(f"Error al crear cliente: {e}")
    client = None
```

## 1. Listar Fondos Disponibles

```python
# Listar fondos
if client:
    try:
        print("Obteniendo lista de FCI...")
        fondos = client.get_fci_list()
        
        if fondos:
            print(f"\nTotal de fondos: {len(fondos)}")
            
            print("\nPrimeros 10 fondos:")
            for i, fondo in enumerate(fondos[:10], 1):
                print(f"  {i}. {fondo.simbolo} - {fondo.nombre}")
                print(f"     Tipo: {fondo.tipo} | Moneda: {fondo.moneda}")
                print(f"     Rendimiento: {fondo.rendimiento:+.2f}%")
        else:
            print("No se encontraron fondos")
    except Exception as e:
        print(f"Error: {e}")
```

```python
# Tipos de fondos
if client:
    try:
        print("Obteniendo tipos de fondos...")
        tipos = client.get_fci_types()
        
        if tipos:
            print(f"\nTipos de fondos disponibles: {len(tipos)}")
            for tipo in tipos:
                print(f"  - {tipo.nombre} ({tipo.descripcion})")
    except Exception as e:
        print(f"Error: {e}")
```

```python
# Administradoras
if client:
    try:
        print("Obteniendo administradoras de FCI...")
        admins = client.get_fci_administrators()
        
        if admins:
            print(f"\nAdministradoras: {len(admins)}")
            for admin in admins[:10]:
                print(f"  - {admin.nombre}")
    except Exception as e:
        print(f"Error: {e}")
```

## 2. Detalle de un Fondo

```python
# Detalle de fondo
if client:
    try:
        # Usar el primer fondo de la lista
        fondos = client.get_fci_list()
        if fondos:
            simbolo = fondos[0].simbolo
            print(f"Obteniendo detalle de {simbolo}...")
            
            detalle = client.get_fci_detail(simbolo)
            
            if detalle:
                print(f"\nFondo: {detalle.nombre}")
                print(f"Simbolo: {detalle.simbolo}")
                print(f"Tipo: {detalle.tipo}")
                print(f"Moneda: {detalle.moneda}")
                print(f"\nValor cuotaparte: ${detalle.valor_cuotaparte:,.4f}")
                print(f"Rendimiento: {detalle.rendimiento:+.2f}%")
                print(f"Patrimonio: ${detalle.patrimonio:,.0f}")
                
                if detalle.descripcion:
                    print(f"\nDescripcion: {detalle.descripcion[:200]}...")
    except Exception as e:
        print(f"Error: {e}")
```

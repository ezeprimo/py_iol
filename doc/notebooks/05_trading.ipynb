## Documentacion de Trading

### Metodos disponibles:

| Metodo | Descripcion |
|--------|-------------|
| `buy()` | Compra de acciones en pesos |
| `sell()` | Venta de acciones en pesos |
| `buy_dollar_bond()` | Compra de bonos en dolares (especie D) |
| `sell_dollar_bond()` | Venta de bonos en dolares (especie D) |
| `cancel_operation()` | Cancelar una operacion pendiente |

### Parametros comunes:
- `simbolo`: Simbolo del instrumento (ej: "GGAL", "AL30D")
- `cantidad`: Cantidad a operar
- `precio`: Precio limite
- `mercado`: Mercado (default: bCBA)
- `plazo`: Plazo de liquidacion (default: t1)
- `validez`: Fecha de validez de la orden

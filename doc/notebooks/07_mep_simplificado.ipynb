# 07 - Dolar MEP Simplificado

Este notebook prueba las funcionalidades de dolar MEP simplificado de la API de IOL.

## Funcionalidades:
- Estimacion de compra de dolares MEP
- Estimacion de venta de dolares MEP
- Validacion de operaciones MEP
- Ejecucion de operaciones MEP

**ADVERTENCIA:** Las operaciones de compra/venta estan COMENTADAS.

**Nota:** Requiere credenciales validas de IOL en el archivo `.env`

## Configuracion Inicial

```python
import sys
import os
from datetime import datetime
from dotenv import load_dotenv

sys.path.insert(0, os.path.abspath('../..'))

from pyIol import (
    IOLClient, IOLAPIError,
    EstimacionMEP, ParametrosMEP, ValidacionMEP, ResultadoMEP
)
print("Librerias importadas correctamente")
print(f"Fecha y hora: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```

```python
load_dotenv('../../.env')
USERNAME = os.getenv('IOL_USERNAME', 'tu_usuario_iol')
PASSWORD = os.getenv('IOL_PASSWORD', 'tu_password_iol')

if USERNAME == "tu_usuario_iol":
    print("ADVERTENCIA: Configura las credenciales en .env")
else:
    print(f"Credenciales configuradas - Usuario: {USERNAME}")
```

```python
try:
    client = IOLClient(USERNAME, PASSWORD)
    print("Cliente IOL creado correctamente")
except Exception as e:
    print(f"Error al crear cliente: {e}")
    client = None
```

## 1. Estimacion de Compra de Dolares MEP

```python
# Estimacion de compra
if client:
    try:
        print("Obteniendo estimacion de compra de USD 100...")
        estimacion = client.get_mep_buy_estimate(monto_dolares=100.0)
        
        if estimacion:
            print(f"\nEstimacion de compra MEP:")
            print(f"  Dolares a recibir: USD {estimacion.dolares:,.2f}")
            print(f"  Pesos a pagar: ${estimacion.pesos:,.2f}")
            print(f"  Tipo de cambio: ${estimacion.tipo_cambio:,.2f}")
            print(f"  Comisiones: ${estimacion.comisiones:,.2f}")
        else:
            print("No se pudo obtener estimacion")
    except Exception as e:
        print(f"Error: {e}")
```

```python
# Estimacion de venta
if client:
    try:
        print("Obteniendo estimacion de venta de USD 100...")
        estimacion = client.get_mep_sell_estimate(monto_dolares=100.0)
        
        if estimacion:
            print(f"\nEstimacion de venta MEP:")
            print(f"  Dolares a entregar: USD {estimacion.dolares:,.2f}")
            print(f"  Pesos a recibir: ${estimacion.pesos:,.2f}")
            print(f"  Tipo de cambio: ${estimacion.tipo_cambio:,.2f}")
    except Exception as e:
        print(f"Error: {e}")
```

## 2. Validacion de Operacion MEP

```python
# Validar operacion (no la ejecuta)
if client:
    try:
        print("Validando operacion de compra de USD 100...")
        validacion = client.validate_mep_operation(
            tipo_operacion='compra',
            monto=100.0
        )
        
        if validacion:
            print(f"\nResultado de validacion:")
            print(f"  Valida: {'Si' if validacion.es_valida else 'No'}")
            if validacion.mensaje:
                print(f"  Mensaje: {validacion.mensaje}")
            if validacion.errores:
                print(f"  Errores: {validacion.errores}")
    except Exception as e:
        print(f"Error: {e}")
```

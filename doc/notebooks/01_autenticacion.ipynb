# 01 - Autenticacion y Perfil

Este notebook prueba las funcionalidades basicas de autenticacion con la API de IOL.

## Funcionalidades:
- Configuracion inicial e imports
- Creacion del cliente IOL
- Prueba de autenticacion
- Obtencion de datos del perfil de usuario

**Nota:** Requiere credenciales validas de IOL en el archivo `.env`

## 1. Configuracion Inicial

```python
# Importar las librerias necesarias
import sys
import os
from datetime import datetime
import json
from dotenv import load_dotenv

# Agregar el directorio raiz al path (estamos en doc/notebooks/)
sys.path.insert(0, os.path.abspath('../..'))

# Importar pyIOL
try:
    from pyIol import IOLClient, IOLAPIError
    print("Librerias importadas correctamente")
except ImportError as e:
    print(f"Error de importacion: {e}")
    raise

print(f"Fecha y hora: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```

```python
# Configuracion de credenciales
load_dotenv('../../.env')

USERNAME = os.getenv('IOL_USERNAME', 'tu_usuario_iol')
PASSWORD = os.getenv('IOL_PASSWORD', 'tu_password_iol')

if USERNAME == "tu_usuario_iol" or PASSWORD == "tu_password_iol":
    print("ADVERTENCIA: Debes configurar las credenciales en el archivo .env")
else:
    print("Credenciales configuradas")
    print(f"Usuario: {USERNAME}")
```

## 2. Prueba de Autenticacion

```python
# Crear una instancia del cliente IOL
try:
    client = IOLClient(USERNAME, PASSWORD)
    print("Cliente IOL creado correctamente")
except Exception as e:
    print(f"Error al crear cliente: {e}")
    client = None
```

```python
# Probar la autenticacion
if client:
    try:
        print("Probando autenticacion...")
        is_authenticated = client.test_authentication()
        
        if is_authenticated:
            print("Autenticacion exitosa!")
            print("El token de acceso se obtuvo correctamente")
        else:
            print("Fallo la autenticacion")
            print("Verifica tus credenciales de IOL")
            
    except IOLAPIError as e:
        print(f"Error de API: {e}")
    except Exception as e:
        print(f"Error inesperado: {e}")
else:
    print("No se pudo probar la autenticacion (cliente no creado)")
```

```python
# Obtener datos del perfil
if client:
    try:
        print("Obteniendo datos del perfil...")
        profile_data = client.get_profile_data()
        
        print("Datos del perfil obtenidos:")
        print(json.dumps(profile_data, indent=2, ensure_ascii=False))
        
        if 'nombre' in profile_data:
            print(f"\nResumen del perfil:")
            print(f"  Nombre: {profile_data.get('nombre', 'N/A')}")
            print(f"  Apellido: {profile_data.get('apellido', 'N/A')}")
            print(f"  Numero de cuenta: {profile_data.get('numeroCuenta', 'N/A')}")
            print(f"  DNI: {profile_data.get('dni', 'N/A')}")
            print(f"  Perfil inversor: {profile_data.get('perfilInversor', 'N/A')}")
        
    except IOLAPIError as e:
        print(f"Error de API al obtener perfil: {e}")
    except Exception as e:
        print(f"Error inesperado al obtener perfil: {e}")
else:
    print("No se pueden obtener datos del perfil (cliente no creado)")
```

## Limpieza

```python
# Cerrar el cliente para liberar recursos
if client:
    try:
        client.close()
        print("Cliente IOL cerrado correctamente")
    except Exception as e:
        print(f"Error al cerrar cliente: {e}")
```
